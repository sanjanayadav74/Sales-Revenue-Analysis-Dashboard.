```python
df = df.dropna()
```

```python
df.describe()
```

```python
df.columns
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df.info()
```

```python
df["Month"] = df["Order Date"].dt.to_period("M")
df["Year"] = df["Order Date"].dt.year
```

```python
df[["Order Date", "Month", "Year"]].head()
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_quantity = df["Quantity"].sum()

print("Total Sales: $", round(total_sales, 2))
print("Total Profit: $", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Quantity Sold:", total_quantity)
```

```python
import matplotlib.pyplot as plt

category_sales = df.groupby("Category")["Sales"].sum().sort_values()

plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum()

plt.figure(figsize=(6,6))
plt.pie(category_profit,
        labels=category_profit.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Profit Distribution by Category")
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(8,5))
region_sales.plot(kind="bar", color="skyblue")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=0)

plt.show()
```

```python
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
top_products.plot(kind="barh", color="green")

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")

plt.show()
```

```python
top_states = df.groupby("State")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
top_states.plot(kind="barh", color="orange")

plt.title("Top 10 States by Sales")
plt.xlabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df["Sales"], df["Profit"], alpha=0.6)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

# Business Insights

- Technology category generated the highest sales.
- The West region contributed significantly to overall sales.
- A small number of products generated a large share of total revenue.
- California and New York were among the highest-performing states.
- Higher sales do not always lead to higher profits because discounts and costs affect profitability.
